Your main algorithm will assign an **order-priority level**.

## Business rules

3. An urgent order worth R3,000 or more is `High Priority`.

## 4.2 Build the Priority Algorithm in Python

**PAs implemented: PA0203 – Use algorithms; PA0201 – Conduct basic programming**

Complete the function below.


````python
def classify_order_priority(status, stock_available, urgent_delivery, order_total):
    # Activity 4.2: Order Priority Algorithm
    if status == "Cancelled":
        return "Do Not Process"

    elif not stock_available:
        return "Stock Hold"

    elif urgent_delivery and order_total >= 3000:
        return "High Priority"

    elif urgent_delivery:
        return "Medium Priority"

    elif order_total >= 3000:
        return "Medium Priority"

    else:
        return "Standard Priority"
    pass
````

### Activity 5.3

**PAs implemented: PA0204 and PA0202**

Write queries to calculate:

1. Number of customers by customer type.
2. Number of orders by sales channel.
3. Completed revenue by branch ID.
4. Average completed order total by branch ID.
5. Customers with more than three completed orders using `HAVING`.


````python
# Complete Activity 5.3 here.

import sqlite3

# Create database connection
conn = sqlite3.connect("smartstore.db")


# Query 1: Number of customers by customer type
query1 = """
SELECT
    customer_type,
    COUNT(*) AS total_customers
FROM customers
GROUP BY customer_type;
"""

result1 = pd.read_sql_query(query1, conn)
display(result1)


# Query 2: Number of orders by sales channel
query2 = """
SELECT
    sales_channel,
    COUNT(*) AS total_orders
FROM orders
GROUP BY sales_channel;
"""

result2 = pd.read_sql_query(query2, conn)
display(result2)


# Query 3: Completed revenue by branch
query3 = """
SELECT
    branch_id,
    ROUND(SUM(order_total), 2) AS completed_revenue
FROM orders
WHERE status = 'Completed'
GROUP BY branch_id
ORDER BY completed_revenue DESC;
"""

result3 = pd.read_sql_query(query3, conn)
display(result3)


# Query 4: Average completed order value by branch
query4 = """
SELECT
    branch_id,
    ROUND(AVG(order_total), 2) AS average_order_value
FROM orders
WHERE status = 'Completed'
GROUP BY branch_id;
"""

result4 = pd.read_sql_query(query4, conn)
display(result4)


# Query 5: Customers with more than three completed orders
query5 = """
SELECT
    customer_id,
    COUNT(order_id) AS number_of_orders
FROM orders
WHERE status = 'Completed'
GROUP BY customer_id
HAVING COUNT(order_id) > 3;
"""

result5 = pd.read_sql_query(query5, conn)
display(result5)

````

## 5.4 SQL Joins

**PA implemented: PA0204 – Work with software packages**

A join combines related tables.

The query below combines orders and branches.


````python
query = '''
SELECT
    o.order_id,
    o.order_date,
    b.branch_name,
    o.order_total
FROM orders AS o
INNER JOIN branches AS b
    ON o.branch_id = b.branch_id
WHERE o.status = 'Completed'
ORDER BY o.order_total DESC
LIMIT 10;
'''

display(pd.read_sql_query(query, connection))
````

Create one SQL query that classifies stock as:

- `Out of Stock` when quantity is 0
- `Low Stock` when quantity is above 0 but at or below reorder level
- `Enough Stock` when quantity is above reorder level

Include the branch name and product name.


````python
# Complete Activity 5.5 here.

# Question 4: Stock status by branch

query_stock_status = """
SELECT
    b.branch_name,
    p.product_name,
    s.quantity_on_hand,
    p.reorder_level,

    CASE
        WHEN s.quantity_on_hand = 0
            THEN 'Out of Stock'

        WHEN s.quantity_on_hand <= p.reorder_level
            THEN 'Low Stock'

        ELSE 'Enough Stock'

    END AS stock_status

FROM stock s

INNER JOIN branches b
    ON s.branch_id = b.branch_id

INNER JOIN products p
    ON s.product_id = p.product_id;
"""

stock_status = pd.read_sql_query(query_stock_status, connection)

display(stock_status)
````
